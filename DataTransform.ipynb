```python
# テキストファイルを読み込む
inputfile = "作品集.txt"
with open(inputfile, 'r', encoding='utf-8') as f:
    # ファイル全体を読み込み、改行ごとに分割してリスト化する
    clean_list2 = f.read().splitlines()
# '' を取り除く
clean_list = [item for item in clean_list2 if item != '']
print(clean_list)
```

```python
# テキストファイルを読み込む
inputfile = "Data_sorted.txt"
with open(inputfile, 'r', encoding='utf-8') as f:
    # ファイル全体を読み込み、改行ごとに分割してリスト化する
    Kunieda_list = f.read().splitlines()

print(Kunieda_list)
```

```python
new_listB = []

for b in Kunieda_list:
    for a in clean_list:
        if a in b:  # listAの文字がlistBの要素に含まれているか
            new_listB.append(b)
            print(a)
            break   # 1つでも見つかったら、重複登録を防ぐためにループを抜ける

#print(new_listB)
```

```python
new_listB
```

```python
# 'w' は書き込みモード（ファイルが存在しなければ新規作成、あれば上書き）
with open('output.txt', 'w', encoding='utf-8') as f:
    for item in new_listB:
        f.write(item + '\n')  # 末尾に改行を付けて書き込む
```

```python
import re

# ==========================================
# 1. toc.txt を読み込み・整形
# ==========================================
with open('toc.txt', 'r', encoding='utf-8') as f:
    toc_lines = f.readlines()

# 先頭2行をスキップ
toc_lines = toc_lines[2:]

clean_toc = []
for line in toc_lines:
    line = line.strip()

    # 空行は除外
    if not line:
        continue

    # 先頭の "- " を削除
    line = re.sub(r'^-\s*', '', line)

    # 末尾の「（XXXX年）」または「（出版年不明）」を削除
    line = re.sub(r'（\d{4}年）$', '', line)
    line = re.sub(r'（出版年不明）$', '', line)

    # 前後空白除去
    line = line.strip()

    clean_toc.append(line)

# 確認用
print("=== toc 整形後 ===")
for x in clean_toc:
    print(x)


# ==========================================
# 2. Data.txt を読み込み
# ==========================================
with open('Data.txt', 'r', encoding='utf-8') as f:
    data_lines = [line.rstrip('\n') for line in f]

# 空行を除きたい場合は以下を使う
# data_lines = [line.rstrip('\n') for line in f if line.strip()]


# ==========================================
# 3. 並べ替えキーを決める関数
# ==========================================
def sort_key(data_line):
    """
    Data.txt の1行について、
    toc.txt のどの行に最初に一致するかを返す。
    一致しない場合は最後尾になる大きな値を返す。
    """
    for idx, toc_item in enumerate(clean_toc):
        if toc_item in data_line:
            return idx
    return len(clean_toc)  # 一致しないものは最後


# ==========================================
# 4. Data.txt の行を並べ替え
# ==========================================
sorted_data_lines = sorted(data_lines, key=sort_key)


# ==========================================
# 5. 結果表示
# ==========================================
print("\n=== 並べ替え後 Data ===")
for line in sorted_data_lines:
    print(line)


# ==========================================
# 6. 必要なら保存
# ==========================================
with open('Data_sorted.txt', 'w', encoding='utf-8') as f:
    for line in sorted_data_lines:
        f.write(line + '\n')
```
